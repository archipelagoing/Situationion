# SituatiONION V7 — Causal Intervention & Predictive Validity

## Central Research Question

> **Does GPT-2 actually use the relational representations identified in V5 and V6, and can the observable properties of those representations predict which information is causally important?**

V5 established **what information is decodable beyond strong controls**.

V6 investigates:

1. how relational accessibility redistributes across depth,
2. whether relational representations generalize compositionally,
3. whether different relations occupy shared or distinct representational geometries.

V7 moves from **observational evidence to causal evidence**.

The central distinction is:

\[
\text{information that is decodable}
\neq
\text{information that is causally used}
\]

The larger goal is to determine whether the properties measured in V5/V6 can distinguish information that GPT-2 merely **contains** from information that GPT-2 actually **computes with**.

---

# Main Hypotheses

## H1 — Control-adjusted decodability predicts causal relevance

Representations with greater **conservative advantage** should produce stronger and more selective causal intervention effects than representations that are highly decodable but fully explained by lexical or random-projection controls.

V5 provides an especially useful contrast:

| Variable | V5 raw decodability | Conservative advantage |
|---|---:|---:|
| Event state | high/perfect | ~0 |
| Polarity | high/perfect | ~0 |
| Role assignment | high/perfect | +0.250 |
| Cause-holder role | high/perfect | +0.333 |
| Temporal relation | high/perfect | +0.500 |

Therefore:

\[
\text{Causal Effect}
\sim
\text{Conservative Advantage}
\]

should potentially outperform:

\[
\text{Causal Effect}
\sim
\text{Raw Decodability}
\]

---

## H2 — Representational migration predicts causal handoff

If V6 finds that relational accessibility redistributes between readouts across depth, then causal influence may redistribute with it.

For a transition:

\[
R_A \rightarrow R_B
\]

the predicted pattern is:

\[
Accessibility_A \downarrow
\qquad
Accessibility_B \uparrow
\]

accompanied by:

\[
CausalInfluence_A \downarrow
\qquad
CausalInfluence_B \uparrow
\]

Evidence for both patterns would support a **computational handoff** interpretation.

Importantly, V6 migration alone does **not** establish that information literally moves through the network. V7 is the experiment that tests whether accessibility changes correspond to changes in causal control.

---

## H3 — Compositionally general representations should remain causally effective

If V6 shows that a relational representation survives difficult unseen entity × verb × template combinations, then interventions on that representation should also produce relation-specific effects on novel compositions.

This tests whether the causal mechanism itself generalizes rather than operating only on familiar lexical patterns.

---

## H4 — Relational geometry predicts causal organization

If V6 finds that some relational variables share representational geometry while others are distinct, causal interventions should respect that organization.

For example, if role and cause representations overlap more strongly than role and temporal representations, interventions may show correspondingly greater causal overlap between role and cause.

This is a hypothesis, not an assumed result.

# V7's Most Important Possible Result

The goal is **not** simply to show that activation patching changes GPT-2's output.

The strongest result would be evidence that:

> **Information can be highly decodable from a language model without being strongly causally relevant, while control-adjusted representational properties and their dynamics predict which information the model actually uses.**

An even stronger result would show that:

> **Representational migration corresponds to a causal handoff: as relational information becomes accessible through a different internal representation, causal control over model behavior shifts with it.**

And the strongest extension would be:

> **A causal-relevance predictor developed from GPT-2 XL's observational representations successfully predicts causal importance prospectively in another model.**

That would move SituatiONION beyond describing GPT-2 XL and toward a more general mechanistic-interpretability question:

# **How can we tell the difference between information an AI merely contains and information it is actually computing with?**
